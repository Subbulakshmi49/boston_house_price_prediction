# import required libraries

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import load_model
from tensorflow.keras.datasets import boston_housing

# Load dataset
```

```python
(X_train, y_train), (X_test, y_test) = boston_housing.load_data()
```

# continue with scalling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

# build ANN model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()

model.add(Dense(64, activation='relu', input_dim=13))
model.add(Dense(32, activation='relu'))
model.add(Dense(1))
```

# compile  model

```python
model.compile(
    optimizer='adam',
    loss='mean_squared_error',
    metrics=['mae']
)
```

# Train model

```python
history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=16,
    validation_split=0.2
)
```

# Evaluate model

```python
loss, mae = model.evaluate(X_test, y_test)

print("Loss:", loss)
print("MAE:", mae)
```

# save model

```python
model.save("boston_house_price_model.h5")

print("Model Saved Successfully")
saved_model = load_model("boston_house_price_model.h5")

print("Saved Model Loaded Successfully")
new_data = np.array([[0.1,
                      18.0,
                      2.31,
                      0,
                      0.5,
                      6.5,
                      65.2,
                      4.09,
                      1,
                      296,
                      15.3,
                      396.9,
                      4.98]])
new_data_scaled = scaler.transform(new_data)
prediction = saved_model.predict(new_data_scaled)

print("Predicted House Price :", prediction[0][0])
```
